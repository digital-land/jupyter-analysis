## Single matches

for single match - if lookup-org-a=government-organisation:PB1164 and look-up-org-b is in ODP, check if entity-a-name and entity-b-name is same/ almost same

redirect entity-a to entity-b in the format below

entity-a,301,entity-b,,redirect old entity single match to LPA entity,today's date,

This should be added to the old-entity.csv in conservation area pipeline

```python
from difflib import SequenceMatcher

df_single= df.loc[(df['message'] == 'single_match') & (df['dataset'] == 'conservation-area')].copy()
df_single = df_single.loc[(df_single['lookup-org-a'] == 'government-organisation:PB1164') & (df_single['in-odp'] == True)]#[['entity_a', 'entity_a_name', 'entity_b', 'entity_b_name']]

# Add a similarity score column
df_single['name_similarity'] = df_single.apply(
    lambda row: SequenceMatcher(None, 
                                str(row['entity_a_name']).lower(), 
                                str(row['entity_b_name']).lower()).ratio() * 100,
    axis=1
)

# Filter for similar names (e.g., > 80% similar)
similar = df_single[df_single['name_similarity'] > 80].copy()
```

```python
similar = similar[['entity_a', 'entity_b']]
similar['status'] = 301
similar['notes'] = 'Redirect old entity single match to LPA entity'
similar['entry-date'] = datetime.now().strftime('%Y-%m-%d')
similar.rename(columns={'entity_a': 'old-entity', 'entity_b': 'entity'}, inplace=True)
```
