## Purpose

Read Newcastle City Council's Tree Preservation Order WFS layer into a GeoDataFrame and export it to a local GeoJSON file.

```python
import geopandas as gpd
```

## Data Import

```python
url = (
    "https://gismo.newcastle.gov.uk/geoserver/gis/ows?service=WFS&version=1.0.0"
    "&request=GetFeature&typeName=gis%3ATPO_OpenData&maxFeatures=9000"
    "&outputFormat=application%2Fjson"
)

gdf = gpd.read_file(url)

print("rows:", len(gdf))
print("crs:", gdf.crs)
print(gdf.columns.tolist())
gdf.head(2)
```

## Export

CRS is left as EPSG:27700 (British National Grid), matching the source data.

```python
out_path = "newcastle_tpo.geojson"
gdf.to_file(out_path, driver="GeoJSON")

print("saved", len(gdf), "features to", out_path)
```
